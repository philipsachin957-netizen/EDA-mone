# CRAMMER'S V
## * It doesn't tell us that one variable causes the other. 
##  * It just measures how strongly their categories are related.

## screening categorical relationships
## calculating for single pair

```python
!pip install scipy
```

```python
from scipy.stats import chi2_contingency
table=pd.crosstab(df_decoded["personal_status_sex"],
                 df_decoded["job"])

chi2,p,dof,expected =chi2_contingency(table)
n=table.sum().sum()

phi2=chi2/n

r,k=table.shape

cramers_v=(phi2/min(k-1,r-1))**0.5

print("Cramers V :", cramers_v)
```

## * cramer's v - how strong is the association
## * p-value -checks statistical evidence for an association exists

```python
from scipy.stats import chi2_contingency 
import math

sensitive_feature="personal_status_sex"

proxy_features=["checking_status",
               "credit_history",
               "purpose",
               "savings",
               "employment",
               "other_debtors",
               "property", 
               "other_installment_plans",
               "housing","job",
               "telephone",
               "foreign_worker"]
for feature in proxy_features:
    table=pd.crosstab(df_decoded[sensitive_feature],
               df_decoded[feature])
    chi2,p,dof,expected=chi2_contingency(table)
    
    
    n=table.sum().sum()
    r,k=table.shape
    
    cramers_v=math.sqrt((phi2/min(k-1,r-1)))
    print(f"{feature:25s} Cramer's V = {cramers_v:.3f}     p-value={p:.4f}")
```

### Think of the p-value as a "how surprising is this?" meter.

### * Lower p-value → the observed association would be more surprising if there were actually no association → stronger statistical evidence of an association.
### * Higher p-value → the observed association is more compatible with no association → weaker evidence

```python
from sklearn.inspection import permutation_importance
result= permutation_importance(model,
                              X_test_original,
                              y_test,
                              n_repeats=10,
                              random_state=42)
print(result)
```

```python
importance_df=pd.DataFrame({"feature":X_test_original.columns,
                           "importance_mean":result.importances_mean,
                           "importance_std":result.importances_std})
importance_df.sort_values("importance_mean",ascending=False)
```

## Random variation/noise in the estimate means:

### Our experiment isn't giving us a very stable measurement of the feature's importance.

## It does NOT necessarily mean:

### The feature itself is noise.

# which features look unimportant - previous step
# does removing them actually acffect the model - current step

```python
p=0
features_to_remove = []
for feature,importance in zip(importance_df["feature"], importance_df["importance_mean"]):
    if importance<=0:
        p+=1
        features_to_remove.append(feature)
print(features_to_remove)
print(p)
```

```python
# how to decide which features to remove?
# we shouldnt arbitrarily remove every feature below a importance value, instead  use the ranking to construct several candidate feature sets and see how well each performs on the validation set.
# for the 22 feature set we can experiment it as 
# CANDIDATE                         FEATURES RETAINED                                            FEATURES REMOVED
# full model                        22                                                           0
# mild reduction                    18                                                           2
# moderate reduction                14                                                           8
# strong reduction                  11                                                           11
# agressive reduction               8                                                            14

# for each candidate we retain the highest ranked features, train a fresh model on the 640 sub training  rows and measure
# validation accruracy and bad-credit recall on the 160 validation rows

# repeat the same procedure for the 20 feature experiment using the proportionally comparable reduction levels.
# before the selection of our condidate , the 200 test rows remain untouched.
```

```python
# candidate numbers of features to retain 
feature_counts_22 =[22,18,14,11,8]

#store ranked feature names
ranked_features_22=importance_22["feature"].tolist()

#display each candidate feature set
for feature_count in feature_counts_22:
    selected_features = ranked_features_22[:feature_count]
    
    print(f"\n Top {feature_count} features:")
    print(selected_features)
    
```

```python
## train and compare the 5 candidates
#Accuracy — overall correct predictions.
#Bad-credit recall — how many actual bad-credit cases we identify.
#Number of features — how much we've reduced the input data.

from sklearn.metrics import accuracy_score, recall_score
import pandas as pd

results_22=[]

for feature_count in feature_counts_22:
    
    #select top ranked features
    selected_features=ranked_features_22[:feature_count]
    X_train_candidate=X22_train_sub[selected_features].copy()
    X_val_candidate=X22_val[selected_features].copy()
    
    #identify feature types for this candidates
    
    categorical_candidate= X_train_candidate.select_dtypes(include=["object","category"]).columns.tolist()
    
    numerical_candidate=X_train_candidate.select_dtypes(include=["number"]).columns.tolist()
    
    #create preprocessing for this feature set
    preprocessor_candidate= ColumnTransformer(transformers=[("cat", OneHotEncoder(handle_unknown ="ignore"),categorical_candidate),
                                                           ("number","passthrough",numerical_candidate)])
    
    #build a fresh model for evry candidate
    model_candidate=Pipeline(steps=[("preprocessor",preprocessor_candidate),
                                    ("classifier",LogisticRegression(max_iter=3000))])
    #Train and predict
    model_candidate.fit(X_train_candidate, y22_train_sub)
    y_val_pred=model_candidate.predict(X_val_candidate)
    
    #store evaluation results
    results_22.append({"feature_count":feature_count,
                      "accuracy":accuracy_score(y22_val,y_val_pred),
                      "bad_credit_recall":recall_score(y22_val,y_val_pred,pos_label=2)})
    
results_22_df=pd.DataFrame(results_22)
print(results_22_df.to_string(index=False))
```
